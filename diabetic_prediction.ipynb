```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv('diabetes.csv')
df
```

```python
col=df.columns
col
```

```python
col_to_fix=[]
for c in col:
    if 0 in df[col].values:
        col_to_fix.append(c)
col_to_fix
        
```

```python
col_to_fix.remove('Outcome')
```

```python
col_to_fix
```

```python
for col in col_to_fix:
    df[col].replace(0,df[col].median())
df
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df.info()
```

```python
x=df.drop('Outcome',axis=1)
x
```

```python
y=df['Outcome']
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,stratify=y)
```

```python
scaler=StandardScaler()
x_scaled_train=scaler.fit_transform(x_train)
x_scaled_test=scaler.transform(x_test)
```

```python
models=[]
models.append(('Linear',LinearRegression()))
models.append(('Nav',GaussianNB()))
models.append(('dec',DecisionTreeClassifier()))
```

```python
for name,model in models:
    model.fit(x_scaled_train,y_train)
```

```python
names=[]
scores=[]

for name,model in models:
    y_unknown_values=model.predict(x_scaled_test)
    names.append(name)
    scores.append(y_unknown_values)
```

```python
df_results = pd.DataFrame({
    'Model': names,
    'Accuracy': scores
})
df_results
```

```python
import pandas as pd

df_preds = pd.DataFrame()

for name, model in models:
    y_pred = model.predict(x_scaled_test)
    df_preds[name] = y_pred   # ✅ each model becomes a column

print(df_preds.head())
```

```python
tuv=LinearRegression()
tuv.fit(x_scaled_train,y_train)
y_prediction=tuv.predict(x_scaled_test)
```
